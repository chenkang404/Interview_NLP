```python
from datasketch import MinHash, MinHashLSH

# 创建两个文本的n-gram集合
text1 = set("apple".encode('utf-8')[i:i+3] for i in range(len("apple")-2))  # 3-gram
text2 = set("apples".encode('utf-8')[i:i+3] for i in range(len("apples")-2))

# 生成MinHash签名（k=100个哈希函数）
m1, m2 = MinHash(num_perm=100), MinHash(num_perm=100)
for s in text1:
    m1.update(s)
for s in text2:
    m2.update(s)

# 创建LSH索引并查询相似性
lsh = MinHashLSH(threshold=0.5, num_perm=100)
lsh.insert("text1", m1)
result = lsh.query(m2)
print("相似文本：", result)  # 若Jaccard相似度>0.5则返回"text1"
```
