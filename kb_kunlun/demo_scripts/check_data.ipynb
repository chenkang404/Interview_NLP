```python
#知识库列表
def query_knowledge_base(url, headers, params):
    try:
        response = requests.get(url, headers=headers, params=params)
        response.raise_for_status()  # 如果响应状态码不是200，将抛出HTTPError异常
        return response.json()  # 返回JSON格式的响应数据
    except requests.exceptions.HTTPError as errh:
        return f"HTTP Error: {errh}"
    except requests.exceptions.ConnectionError as errc:
        return f"Error Connecting: {errc}"
    except requests.exceptions.Timeout as errt:
        return f"Timeout Error: {errt}"
    except requests.exceptions.RequestException as err:
        return f"OOps: Something Else: {err}"

# 使用函数
url = 'http://172.16.2.77:56966/v1/datasets'
headers = {
    'Authorization': 'Bearer dataset-F8zSYohCngBjVNsh2NGBv5Bl'  # 请替换为您的实际Token
}
params = {
    'page': 1,
    'limit': 20
}

knowledge_base_data = query_knowledge_base(url, headers, params)
print(knowledge_base_data)
print(len(knowledge_base_data["data"]))
for one_kb in knowledge_base_data["data"]:
    print(one_kb["id"])
    print(one_kb["name"])

# knowledge_base_data["data"][0]
```

```python

def create_dataset(api_key, name, permission="only_me"):
    """
    创建一个新的知识库。

    参数:
    - api_key (str): API 密钥。
    - name (str): 数据集的名称。
    - permission (str): 数据集的权限，默认为 "only_me"。

    返回:
    - dict: API 的响应结果。
    """
    # 设置请求 URL 和头部
    url = "http://172.16.2.77:56966/v1/datasets"
    headers = {
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json"
    }

    # 请求体
    payload = {
        "name": name,
        "permission": permission
    }

    # 发送 POST 请求
    response = requests.post(url, headers=headers, data=json.dumps(payload))

    # 返回解析后的响应
    if response.status_code == 200:
        return response.json()
    else:
        return {"error": response.status_code, "message": response.text}

# 使用示例
api_key = "dataset-F8zSYohCngBjVNsh2NGBv5Bl"  # 替换为实际 API 密钥
name = "昆仑健康核保"  # 数据集名称
permission = "only_me"  # 可选，默认为 "only_me"

result = create_dataset(api_key, name, permission)
print(result)
```

```python
# 通过文本创建文档

def create_document_with_text(dataset_id, api_key, name, text, indexing_technique="high_quality", mode="automatic"):
    """
    创建文档到指定的数据集。

    参数:
        dataset_id (str): 数据集的ID。
        api_key (str): API密钥。
        name (str): 文档名称。
        text (str): 文档内容。
        indexing_technique (str): 索引技术，默认值为 "high_quality"。
        mode (str): 处理规则模式，默认值为 "automatic"。

    返回:
        dict: 响应结果。
    """
    url = f"http://172.16.2.77:56966/v1/datasets/{dataset_id}/document/create-by-text"
    headers = {
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json"
    }
    payload = {
        "name": name,
        "text": text,
        "indexing_technique": indexing_technique,
        'process_rule': {
                'rules': {
                    'pre_processing_rules': [
                        # {'id': 'remove_extra_spaces', 'enabled': True},
                        {'id': 'remove_urls_emails', 'enabled': True}
                    ],
                    'segmentation': {
                        'separator': '\n\n',
                        'max_tokens': 1000
                    }
                },
                'mode': 'custom'
            }
    }

    response = requests.post(url, headers=headers, json=payload)

    if response.status_code == 200:
        return response.json()  # 成功时返回 JSON 响应
    else:
        return {"status_code": response.status_code, "error": response.text}  # 失败时返回错误信息

# 示例调用
dataset_id = "c02c5910-993a-4bed-8985-c7f23572b837"
api_key = "dataset-F8zSYohCngBjVNsh2NGBv5Bl"
name = "中再疾病函数对照表"
text = ""
df_dise_2_keyword = pd.read_excel("./中再疾病函数对照表.xlsx",keep_default_na=False)
disease_n_2_list = {}
for idx,row in df_dise_2_keyword.iterrows():
    disease_n = row["disease_n"].strip()
    disease_list = eval(row["disease_list"])
    # print(type(disease_list))
    try:
        if disease_n  not in disease_n_2_list and disease_n!="" and disease_n!="转人工":
            disease_n_2_list[disease_n] = []
        disease_n_2_list[disease_n].extend(disease_list)
    except Exception as es:
        print(es)

text_list = [json.dumps({"disease_n":k,"disease_keyword":v},ensure_ascii=False) for k,v in disease_n_2_list.items()]
text = "\n\n".join(text_list)
result = create_document_with_text(dataset_id, api_key, name, text)
print(result)

```

```python
#删除文档
def delete_document(dataset_id, document_id, api_key):
    """
    删除指定数据集中的文档。

    参数:
    - dataset_id (str): 数据集 ID。
    - document_id (str): 要删除的文档 ID。
    - api_key (str): API 密钥。

    返回:
    - dict: API 的响应结果。
    """
    # 设置请求 URL 和头部
    url = f"http://172.16.2.77:56966/v1/datasets/{dataset_id}/documents/{document_id}"
    headers = {
        "Authorization": f"Bearer {api_key}"
    }

    # 发送 DELETE 请求
    response = requests.delete(url, headers=headers)

    # 返回解析后的响应
    if response.status_code == 200:
        return {"message": "Document deleted successfully"}
    else:
        return {"error": response.status_code, "message": response.text}

# 使用示例
dataset_id = "88a91412-da40-4892-b274-d9afe4880d2f"  # 替换为实际数据集 ID
document_id = "78dccd90-afee-49c8-b80e-14bd3382ce8d"  # 替换为实际文档 ID
api_key = "dataset-F8zSYohCngBjVNsh2NGBv5Bl"  # 替换为实际 API 密钥

result = delete_document(dataset_id, document_id, api_key)
print(result)
```

# 检索数据库

```python
import requests
import json

def retrieve_documents(dataset_id, api_key, query, search_method="semantic_search", top_k=3):
    """
    从数据集中检索文档。

    参数:
    - dataset_id (str): 数据集 ID。
    - api_key (str): API 密钥。
    - query (str): 查询关键词。
    - search_method (str): 检索方法，默认为 "keyword_search"。
    - top_k (int): 返回的结果数量，默认为 1。

    返回:
    - dict: API 的响应结果。
    """
    # 设置请求 URL 和头部
    url = f"http://172.16.2.77:56966/v1/datasets/{dataset_id}/retrieve"
    headers = {
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json"
    }

    # 定义请求体
    payload = {
        "query": query,
        "retrieval_model": {
            "search_method": search_method,
            "reranking_enable": False,
            "reranking_mode": None,
            "reranking_model": {
                "reranking_provider_name": "",
                "reranking_model_name": ""
            },
            "weights": None,
            "top_k": top_k,
            "score_threshold_enabled": False,
            "score_threshold": None
        }
    }

    # 发送 POST 请求
    response = requests.post(url, headers=headers, data=json.dumps(payload))

    # 返回解析后的响应
    if response.status_code == 200:
        return response.json()
    else:
        return {"error": response.status_code, "message": response.text}

# 使用示例
# dataset_id = "88a91412-da40-4892-b274-d9afe4880d2f"  # 替换为实际数据集 ID
dataset_id = "473a6e55-7d68-4f9f-9de4-f6213ce93ae0"  # 替换为实际数据集 ID

api_key = "dataset-F8zSYohCngBjVNsh2NGBv5Bl"  # 替换为实际 API 密钥
query = "甲状腺炎"

result = retrieve_documents(dataset_id, api_key, query)
print(result)
```

# 对昆仑健康_核保指引.txt进行数据处理

```python
with open("./昆仑健康_核保指引.txt","r")as f:
    datas = f.readlines()
```

```python
one_dict = eval(datas[0].strip())
print(type(one_dict))
colums_ = one_dict.keys()
print(colums_)
print(one_dict)
```

```python
good_datas = []
bad_datas = []
for one_data in datas:
    one_data = one_data.strip()
    if one_data == "":
        continue
    dict_data = eval(one_data)
    all_good = True
    for key in dict_data.keys():
        if key not in colums_:
            all_good = False
    if all_good:
        good_datas.append(dict_data)
    else:
        bad_datas.append(dict_data)
```
