```python
import pandas as pd
```

```python
file_path = './昆仑健康_核保指引_20230522.xlsx'
excel_data = pd.ExcelFile(file_path)
```

```python
excel_data.sheet_names
```

```python
sheet_name = "妇科肿物"
sheet_data = pd.read_excel(excel_data, sheet_name=sheet_name,keep_default_na=False)
```

```python
sheet_data.columns
```

```python
have_value = ""
for idx,row in sheet_data.iterrows():
    if idx == 2:
        continue
    updata_name = "检查结果"
    if row[updata_name]!="":
        have_value = row[updata_name]
    else:
        sheet_data.loc[idx,updata_name] = have_value
```

```python
sheet_data["检查结果"] = sheet_data["检查结果"] + "," +sheet_data["Unnamed: 4"]
```

```python
sheet_data["疾病"] = sheet_data["疾病"] + "/" + sheet_data["Unnamed: 1"]
```

```python
sheet_data.drop(["Unnamed: 1"],axis=1,inplace=True)
```

```python
for idx,row in sheet_data.iterrows():
    if row["防癌"]=="":
        sheet_data.loc[idx,"防癌"] = row["重疾"]
    if row["意外险"]=="":
        sheet_data.loc[idx,"意外险"] = row["重疾"]
    if row["护理险"]=="":
        sheet_data.loc[idx,"护理险"] = row["重疾"]
    if row["医疗险"]=="":
        sheet_data.loc[idx,"医疗险"] = row["重疾"]
```

```python
sheet_data.head(20)
```

```python
with open("./save_data/妇科肿物.txt","w") as f:
    for idx,row in sheet_data.iterrows():
        item_ = row.to_json(force_ascii=False)
        f.write(item_ + '\n\n')
```

```python
df_conclusions = pd.read_csv("./utils/disease_2_conclusion_v3.csv",keep_default_na=False)
```

```python
df_conclusions_v2 = df_conclusions[df_conclusions["file_name"]!="妇科肿物.txt"]
```

```python
df_conclusions_v2.head()
```

```python
data_save = []
for idx,row in sheet_data.iterrows():
    item_ = row.to_json(force_ascii=False)
    dict_data = {
        "disease_name":row["疾病"],
        "file_name":"妇科肿物.txt",
        "conclusion":item_
    }
    data_save.append(dict_data)
```

```python
df_conclusions_save = pd.concat([df_conclusions_v2,pd.DataFrame(data_save)],ignore_index=True)
```
