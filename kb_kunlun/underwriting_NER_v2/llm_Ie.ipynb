````python
import requests
import json


def qa_baichuan(input_):
    # 定义API的URL
    url = "https://api.baichuan-ai.com/v1/chat/completions"
    baichuan_key = "sk-4ce3bfe548d2bd5209aee1e19f72f4b8"

    # 定义请求头
    headers = {
        "Content-Type": "application/json",
        "Authorization": f"Bearer {baichuan_key}"  # 请确保你已经定义了API_KEY变量
    }

    # 定义请求体
    data = {
        "model": "Baichuan4",
        "messages":[
            {"role": "user",
        "content": input_}
        ],
        "temperature": 0.3,
        "stream": False
    }

    # 发送POST请求
    # response = requests.post(url, headers=headers, data=json.dumps(data))
    response = requests.post(url, headers=headers, json=data)


    # 检查响应状态码
    if response.status_code == 200:
        # 如果请求成功，打印响应内容
        # print(response.json())
        return response.json()["choices"][0]["message"]["content"]
    else:
        # 如果请求失败，打印错误信息
        print(f"Request failed with status code {response.status_code}")
````

````python
def match_struct_data(gen_result_bc):
    pattern = r"```python(.*?)```"
    matches = re.findall(pattern, gen_result_bc, re.DOTALL)
    if matches:
        code_content = matches[0].strip()
        code_dict = eval(code_content)
        # print("type:",type(code_dict))
        # print(code_dict)
    else:
        print("not match")
        code_dict = {}
    return code_dict
````

````python
# match_struct_data(gen_result_bc)
save_2_xlsx = {}
for key,value in save_result.items():
    print(key)
    df_list = []
    for dix, one_v in enumerate(value):
        # print(one_v)
        description = one_v["input_description"]
        save_row = {"字段":f"测试样例:{dix+1}","百川结果":""}
        df_list.append(save_row)

        save_row = {"字段":description,"百川结果":""}
        df_list.append(save_row)
        gen_rel = one_v["baichuan_gen"]
        dict_gen_rel = match_struct_data(gen_rel)
        print(type(dict_gen_rel))
        def recursion_(dict_data):
            for k,v in dict_data.items():
                v_type = type(v)
                if v_type == type({}):
                    save_row =  {"字段":k,"百川结果":""}
                    df_list.append(save_row)
                    recursion_(v)
                elif v_type == type([]):
                    save_row =  {"字段":k,"百川结果":""}
                    df_list.append(save_row)
                    for one_sub in v:
                        for k,v in one_sub.items():
                            save_row =  {"字段":k,"百川结果":v}
                            df_list.append(save_row)
                else:
                    save_row =  {"字段":k,"百川结果":v}
                    df_list.append(save_row)
        recursion_(dict_gen_rel)
    df_2_save = pd.DataFrame(df_list)
    save_2_xlsx[key] = df_2_save
````
