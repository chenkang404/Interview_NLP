```python
# 1. 编辑距离（Edit Distance）
def min_edit_distance(word1, word2):
    m, n = len(word1), len(word2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    
    for i in range(m + 1):
        for j in range(n + 1):
            if i == 0:
                dp[i][j] = j
            elif j == 0:
                dp[i][j] = i
            elif word1[i - 1] == word2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(dp[i - 1][j], dp[i][j - 1], dp[i - 1][j - 1])
    
    return dp[m][n]

# 2. 最长回文子序列
def longest_palindromic_subsequence(s):
    n = len(s)
    dp = [[0] * n for _ in range(n)]
    
    for i in range(n):
        dp[i][i] = 1
    
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            if s[i] == s[j]:
                dp[i][j] = dp[i + 1][j - 1] + 2
            else:
                dp[i][j] = max(dp[i + 1][j], dp[i][j - 1])
    
    return dp[0][n - 1]

# 3. 最长公共子序列（LCS）
def longest_common_subsequence(text1, text2):
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i - 1] == text2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])
    
    return dp[m][n]

# 4. 最长递增子序列（LIS）
def length_of_lis(nums):
    if not nums:
        return 0
    dp = [1] * len(nums)
    for i in range(len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)

# 5. 最大子序和
def max_subarray(nums):
    current_sum = max_sum = nums[0]
    for num in nums[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
    return max_sum

# 6. 背包问题（0/1 Knapsack）
def knapsack(weights, values, capacity):
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    
    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(dp[i - 1][w], dp[i - 1][w - weights[i - 1]] + values[i - 1])
            else:
                dp[i][w] = dp[i - 1][w]
    
    return dp[n][capacity]

# 测试所有函数
if __name__ == "__main__":
    # 编辑距离
    print("编辑距离:", min_edit_distance("horse", "ros"))  # 输出: 3
    
    # 最长回文子序列
    print("最长回文子序列:", longest_palindromic_subsequence("bbbab"))  # 输出: 4
    
    # 最长公共子序列
    print("最长公共子序列:", longest_common_subsequence("abcde", "ace"))  # 输出: 3
    
    # 最长递增子序列
    print("最长递增子序列:", length_of_lis([10, 9, 2, 5, 3, 7, 101, 18]))  # 输出: 4
    
    # 最大子序和
    print("最大子序和:", max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))  # 输出: 6
    
    # 背包问题
    print("背包问题:", knapsack([1, 2, 3], [6, 10, 12], 5))  # 输出: 22
```

```python
def longest_palindromic_substring(s):
    """
    最长回文子串(substring)
    定义：最长回文子串是字符串中连续的一段子串，且这段子串是一个回文，并且长度最长。
    关键点：必须是连续的。
    """
    n = len(s)
    dp = [[False] * n for _ in range(n)]
    start, max_length = 0, 1
    
    for i in range(n):
        dp[i][i] = True  # 单个字符是回文
    
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1
            if s[i] == s[j]:
                if length == 2:  # 两个字符相等
                    dp[i][j] = True
                else:
                    dp[i][j] = dp[i + 1][j - 1]
                if dp[i][j] and length > max_length:
                    start, max_length = i, length
    
    return s[start:start + max_length]

# 测试
print("最长回文子串:", longest_palindromic_substring("bbbab"))  # 输出: "bbb"
```

```python
def max_palindromic_substring(x):
    n = len(x)
    dp = [[False]*n for _ in range(n)]
    start, max_len = 0,1
    for length in range(2,n+1):
        for i in range(n-length+1):
            j = i+length -1
            if x[i] == x[j]:
                if length==2:
                    dp[i][j] = True
                else:
                    dp[i][j] = dp[i+1][j-1]
                if dp[i][j] and length>max_len:
                    start = i
                    max_len = length

    return x[start:start+length]
                

print("最长回文子串:", longest_palindromic_substring("abbadda"))  # 输出: "bbb"
```
